# Zielsetzung
Ziel dieser Arbeit ist die Entwicklung eines Klassifikators für die Erkennung von Emotionen in menschlichen Gesichtern. Dazu sollen Evidenztheorie und Case-Based-Reasoning verwendet werden. Beide Konzepte werden zunächst mit Beispielen erläutert, anschließend wird die Datenbasis vorgestellt und abschließend das Modell trainiert, getestet und evaluiert.

## (i) Evidenztheorie
Beim maschinellen Lernen beschriebt die Evidenztheorie die Theorie der
Glaubensfunktionen (auch Dempster-Shafer-Theorie). Sie arbeitet mit "belief" und "function theory". Sie hilft die Unsicherheit eines Modells zu beschreiben, wenn die Datenbasis unvollständig oder widersprüchlich ist.

---
Grundidee
Statt jeder Hypothese eine einzelne Wahrscheinlichkeit zuzuweisen, verteilt man “Masse” (basic probability assignment) auf Mengen von Hypothesen. Daraus ergeben sich zwei Größen:

Belief (Bel): Wie stark ist eine Hypothese durch die Evidenz mindestens gestützt?
Plausibility (Pl): Wie stark ist sie höchstens mit der Evidenz vereinbar?

Das Intervall [Bel, Pl] drückt aus, wie viel man nicht weiß. Das ist der Hauptunterschied zur klassischen Wahrscheinlichkeitstheorie, die Unwissenheit und Gleichverteilung nicht auseinanderhalten kann.

Typische Anwendungen

Sensorfusion und Klassifikatorfusion: Mit der Dempsterschen Kombinationsregel lassen sich Aussagen mehrerer unabhängiger Quellen zusammenführen.
Unsicherheitsschätzung in neuronalen Netzen: Beim Evidential Deep Learning (z. B. Sensoy et al., 2018) gibt das Netz keine Softmax-Wahrscheinlichkeiten aus, sondern Parameter einer Dirichlet-Verteilung. Daraus lässt sich ableiten, wie sicher oder unsicher das Modell ist, etwa bei Out-of-Distribution-Daten.
Entscheidungen unter Unsicherheit: Medizinische Diagnose, Objekterkennung, Risikoanalyse.
